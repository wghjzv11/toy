## Black-Scholes-Merton Equations
### S를 기초자산으로 하는 파생상품의 가격을 f라고 하자. 
### $f=G(S, t)$
### By Ito's Lemma, 
### $ \Delta f = (\mu S \frac{\partial G }{\partial S} + \frac{\partial G}{\partial t} + \frac{1}{2} \sigma^2 S^2 \frac{\partial^2 G}{\partial S^2}) \Delta t + (\frac{\partial G}{\partial S} \sigma S) \Delta z $
### $ \Delta S = \mu S \Delta t + \sigma S \Delta z $

### $ \prod = f - \frac{\partial G}{\partial S}S $ is a riskless portfolio
### thus it must increment by the risk-free interest rates so that
### $ \Delta \prod = r \prod  t$ 
$$ \left( \frac{\partial f}{\partial t} + \frac{1}{2} \frac{\partial^2 f}{\partial S^2} \sigma^2 S^2 \right) \Delta t = r \left( f - \frac{\partial f}{\partial S} S \right) \Delta t $$

$$ \frac{\partial f}{\partial t} + rS \frac{\partial f}{\partial S} + \frac{1}{2} \sigma^2 S^2 \frac{\partial^2 f}{\partial S^2} = rf  \quad \text{(Black-Scholes)} \quad$$
### Any tradable derivative must satisfy this equation.

15.5 

A stock price is currently $40. Assume that the expected return from the stock is 15%
and that its volatility is 25%. What is the probability distribution for the rate of return
(with continuous compounding) earned over a 2-year period?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean_ret = (0.15 - 0.5 * 0.25 **2) * 2
std_ret = 0.25 * np.sqrt(2)

x = np.linspace(mean_ret - 4*std_ret, mean_ret + 4*std_ret, 100)
y = norm.pdf(x, mean_ret, std_ret)

# 4. 시각화
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', lw=2, label=f'Probability Density Function\n(Mean={mean_ret:.4f}, Std={std_ret:.4f})')
plt.fill_between(x, y, color='skyblue', alpha=0.4)

# 평균 위치 표시
plt.axvline(mean_ret, color='red', linestyle='--', label=f'Mean: {mean_ret:.4f}')

# 그래프 설정
plt.title('Probability Distribution of Rate of Return (2-year period)', fontsize=14)
plt.xlabel('Continuous Compounding Rate of Return', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

15.6

A stock price follows geometric Brownian motion with an expected return of 16% and a
volatility of 35%. The current price is $38.

(a) What is the probability that a European call option on the stock with an exercise
price of $40 and a maturity date in 6 months will be exercised?

(b) What is the probability that a European put option on the stock with the same
exercise price and maturity will be exercised?

```python
mean_lnS = np.log(38)+(0.16 - 0.5*0.35 **2)*0.5
std_lnS = 0.35*np.sqrt(0.5)

# Answer to A
ans_a = norm.cdf(np.log(40), mean_lnS, std_lnS)
print(f"Answer to A: {ans_a}")

# Answer to B
ans_b = 1 - norm.cdf(np.log(40), mean_lnS, std_lnS)
print(f"Answer to A: {ans_b}")
```

```python
import sympy as sp

# 수식을 LaTeX로 예쁘게 출력하기 위한 설정
sp.init_printing()

# 기호 정의 (S, t, r, sigma 등을 변수로 선언)
f, S, t, r, sigma = sp.symbols('f S t r sigma')

# f를 S와 t에 대한 함수로 정의 (f(S, t))
f = sp.Function('f')(S, t)

# 블랙-숄즈 방정식 정의
bs_eq = sp.Equality(
    sp.diff(f, t) + r*S*sp.diff(f, S) + (sp.Rational(1, 2) * sigma**2 * S**2 * sp.diff(f, S, S)), 
    r*f
)

# 출력
bs_eq
```

```python
import pandas as pd 

# 1. 콜론(:)을 사용하여 딕셔너리 생성
a = {"S": 45, "K": 45, "r": 0.05}

# 2. .copy()를 사용하여 독립된 객체로 할당
a1 = a.copy()
a2 = a.copy()
a3 = a.copy()

# 3. 이제 각자 다른 값을 가질 수 있습니다.
a1["T"], a2["T"], a3["T"] = 3/12, 6/12, 12/12
a1['price'], a2['price'], a3['price'] = 7.0, 8.3, 10.5

# 1. 콜론(:)을 사용하여 딕셔너리 생성
b = {"S": 45, "K": 50, "r": 0.05}

# 2. .copy()를 사용하여 독립된 객체로 할당
b1 = b.copy()
b2 = b.copy()
b3 = b.copy()

# 3. 이제 각자 다른 값을 가질 수 있습니다.
b1["T"], b2["T"], b3["T"] = 3/12, 6/12, 12/12
b1['price'], b2['price'], b3['price'] = 3.7, 5.2, 7.5


# 1. 콜론(:)을 사용하여 딕셔너리 생성
c = {"S": 45, "K": 55, "r": 0.05}

# 2. .copy()를 사용하여 독립된 객체로 할당
c1 = c.copy()
c2 = c.copy()
c3 = c.copy()

# 3. 이제 각자 다른 값을 가질 수 있습니다.
c1["T"], c2["T"], c3["T"] = 3/12, 6/12, 12/12
c1['price'], c2['price'], c3['price'] = 1.6, 2.9, 5.1



options = [a1, a2, a3, b1, b2, b3, c1, c2, c3]

results = np.array([bsm.get_implied_vol(**option) for option in options]).reshape(3, 3)

# 1. 먼저 리스트를 데이터프레임으로 변환 (내재변동성 결과가 담긴 상태라고 가정)
# 결과 예시: [{'K': 45, 'T': 0.25, 'iv': 0.757}, ...]
results_df = pd.DataFrame(results) 

# 2. pivot 기능을 사용하여 K는 행으로, T는 열로 보냅니다.
vol_surface = results_df.pivot(index='K', columns='T', values='iv')

print(vol_surface)
```

```python
import pandas as pd
import numpy as np

# ... (기존 변수 설정 부분은 동일하다고 가정) ...

# 1. 계산 결과와 원본 데이터를 합쳐서 리스트 생성
final_data = []
for option in options:
    # bsm 계산 수행 (결과값이 iv라고 가정)
    iv = bsm.get_implied_vol(**option)
    
    # 원본 데이터 딕셔너리에 iv 값을 추가하여 새 리스트에 보관
    item = option.copy()
    item['iv'] = iv
    final_data.append(item)

# 2. 리스트를 바로 DataFrame으로 변환 (K, T, S, r, price, iv 컬럼이 생김)
results_df = pd.DataFrame(final_data)

# 3. pivot을 사용하여 표 만들기
# index(세로)는 행사가 'K', columns(가로)는 만기 'T', 값은 'iv'
vol_surface = results_df.pivot(index='K', columns='T', values='iv')

print(vol_surface)
```

```python
import numpy as np

# 1. 입력된 값 계산
mean_lnS = np.log(50) + (0.18 - 0.5 * 0.30**2) * 2
std_lnS = 0.30 * np.sqrt(2)

# 2. ln(S)의 95% 신뢰구간 (Z = 1.96)
ci_lower_ln = mean_lnS - 1.96 * std_lnS
ci_upper_ln = mean_lnS + 1.96 * std_lnS

# 3. 실제 주가 S의 95% 신뢰구간으로 변환 (Exponential)
ci_lower_S = np.exp(ci_lower_ln)
ci_upper_S = np.exp(ci_upper_ln)

print(f"ln(S)의 평균: {mean_lnS:.4f}")
print(f"S의 95% 신뢰구간: [{ci_lower_S:.2f}, {ci_upper_S:.2f}]")
```

```python
import numpy as np
import pandas as pd
from scipy import stats

data = [30.2, 32.0, 31.1, 30.1, 30.2,
        30.3, 30.6, 33.0, 32.9, 33.0,
       33.5, 33.5, 33.7, 33.5, 33.2]

df = pd.DataFrame({'price': data})
df['return'] = df['price'].pct_change()
df
```

```python
# 로그 수익률 계산
df['log_return'] = np.log(1 + df['data'].pct_change())

# 주간 변동성 (표준편차)
weekly_vol = df['log_return'].std() # 여기서 n-1이 적용됩니다.

# 연 변동성 환산 (주간 -> 연간)
annual_vol = weekly_vol * np.sqrt(52)

print(f"주간 변동성: {weekly_vol:.4f}")
print(f"연 변동성: {annual_vol:.4f}")
```
